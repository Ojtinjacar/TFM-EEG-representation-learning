```python
def plot_window(arr):
    fig = go.Figure()
    N, M = arr.shape
    x = np.arange(arr.shape[1])

    for i in range(arr.shape[0]):
        y = arr[i]
        ch_name = CH_NAMES[i] if N == len(CH_NAMES) else str(i)
        fig.add_trace(
            go.Scatter(
                x=x,
                y=y,
                mode="lines",
                name=ch_name,  # legend shows the id
                customdata=np.full(M, i),
                hovertemplate=f"id={ch_name}<br>x=%{{x}}<br>y=%{{y}}<extra></extra>",
                line=dict(color=f"rgb({colors[i][0]*255}, {colors[i][1]*255}, {colors[i][2]*255})")
            )
        )

    fig.update_layout(xaxis_title="Index", yaxis_title="Value")
    return fig
```
